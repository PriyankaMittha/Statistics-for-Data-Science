```python
df = sns.load_dataset('healthexp')
```

```python
df
```

# covariance

```python
df.cov(numeric_only=True)
```

# co-relation

```python
df.corr(numeric_only=True, method= "pearson")
```

```python
df.corr(numeric_only=True, method= "spearman")
```
